3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
# Code here
import matplotlib.pyplot as plt 
plt.boxplot(df['income'], vert=False)
plt.show()
plt.boxplot(df['education'], vert=False)
plt.show()
plt.boxplot(df['prestige'], vert=False)
plt.show()

import numpy as np

def fns(column):
    min = np.min(df[column])
    q1 = np.percentile(df[column], 25)
    med = np.median(df[column])
    mean = np.mean(df[column])
    q3 = np.percentile(df[column], 75)
    max = np.max(df[column])

    print(f'------------ FNS for {column.capitalize()} -------------')
    print(f"Min: {min}")
    print(f"Q1: {q1}")
    print(f"Median: {med}")
    print(f"Q3: {q3}")
    print(f"Max: {max}")

fns('income')
fns('education')
fns('prestige')
```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
# Code here
standard_dev = df['income'].std()
print()
print(f"Standard Deviation:  {standard_dev:.2f}")

variance = standard_dev ** 2 
print(f"Variance:            {variance:.2f}")
```
