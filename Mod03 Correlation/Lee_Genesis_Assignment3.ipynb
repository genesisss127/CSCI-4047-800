3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *education*

```python
# Code here
correlation, p_value = stats.pearsonr(df["income"], df["education"])

print("Pearson Correlation Coefficient:", correlation)
print("P-value:", p_value)
```

6. Pearson's correlation coefficient OR covariance can both be used to measure the direction of the relationship between two variables (e.g., education and income). Explain why Pearson's correlation coefficient may be better than covariance.

## 6 
The reason Pearson's correlation coefficient is better than covariance is because of its fixed scale of -1.0 to +1.0. This is better because Pearson's correlation coefficient can tell you both the direction and strength of the correlation because of the standardization of the range for the correlation coefficient. For example a negative number leading to a negative direction, and positive number leading to a positive direction of growth. Also with Pearson's correlation coefficient you can test the reliability of the test based on the p-value. However covariance only gives the direction, but not the strength, or the p-value/statistical significance. The reason you cannot determine this reliably from covariance is because it is very dependent on the range of the data. If you change the range from a smaller to bigger range the covariance will change. Pearson's correlation fixes this by dividing the covariance by the standard deviation which gets rid of the issue with different ranges/units.
